```python
def agente_termostato(temperatura_actual, temperatura_objetivo = 22, margen = 1.0 ):
    """ Percepcion (S) :  Terperatura ideal
        Accciones (A) :  Encender_aire_acondicionado, Encender_calefaccion,Apagar_sistema
        Entorno (E) : Habitación cerrada
        Desempeño (P) : Confort termico continuo , ahorro eficiente de energía """

    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen :
        return "Encender_aire_acondionado";
    elif diferencia < -margen:
        return "Encender_calefaccion";
    else :
        return "Apagar_sistema"
```

```python
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5, 23.0, 21.0]
```

```python
for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepcion: {t} C -> Accion del agente: {accion}")
```
